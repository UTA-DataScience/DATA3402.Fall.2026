Why bother? The Central Limit Theorem says grades on one assessment, from one population of students, should come out looking roughly normal. This distribution doesn't — which is a clue, not just noise to smooth over. It's more likely a sign that several populations of students got merged into one dataset: some who didn't take the exam sitting at zero, a normal distribution for the students who did, and possibly a further split among those. Before computing any statistic, it's worth looking at the histogram and asking what population it actually describes.

Taking students out of the data deserves a moment of thought. Decisions about outliers are very context dependent: ask what you're trying to achieve, what assumptions you're making, and whether the outlier breaks them. Here the goal is to grade the students who took the exam. The students at zero, who didn't take it or dropped the class, are a different population and shouldn't influence how everyone else is graded; it wouldn't make sense for the grade boundaries to move just because half the class dropped. That is a decision for this dataset in this context, not a rule. You can't just make a plot and declare some points outliers, because in another problem those points may be the very thing you're looking for.

The mean/standard deviation is being pulled down/wide by the low grades, lets take them out using selection. Note that applying conditionals to DataFrames and Series give you back a boolean Series:

```python
min_exam_1=55
df_0["Exam 1 Fixed"]>min_exam_1
```

```python
exam_1_df=df_0[df_0["Exam 1 Fixed"]>min_exam_1]
```

```python
exam_1_df
```

This is a mistake worth watching for, in your own work and in other people's: computing a mean and standard deviation without first checking whether the data looks like one population. It's easy to hand someone a number like "the mean is 60, sigma is 30" and have it sound authoritative, but a number computed over the wrong population isn't meaningful just because it's a number. Look at the distribution before you trust the statistic.

There's a further point in the mean and standard deviation themselves. Those two numbers completely describe a distribution only if it is normal, so quoting them as the summary of your data quietly assumes that it is. For something that is, say, uniformly distributed, they don't characterize it. You have to look at the histogram and ask whether it looks normal at all. Even with the zeros removed, this one doesn't really look like a bell curve, so these two numbers give a useful sense of what's going on but can't be the whole basis for grading.

Now we can write the function:

```python
def grade_summary(df_0,grades,min_grade=0.):
    df=df_0[df_0>min_grade]
    N_total=sum(df>min_grade)
    hist,bins=np.histogram(df,bins=list(grades.values()))

    print("Mean:",np.mean(df))
    print("Std:",np.std(df))
    
    for grade_letter,bin_count in zip(grades.keys(),hist):
        print(grade_letter+": "+str(bin_count),
              float(bin_count/N_total) )
    
```

How about something that implements a bell curve?

`compute_curve` below takes the grades and an optional minimum grade. If you don't give one, it uses half of the maximum grade, a simple way to cut out the zeros. After the cut it computes the mean and standard deviation, and those give the grade edges directly: B starts at the mean, C one standard deviation below it and D two below, while A starts one standard deviation above the mean but never above the maximum grade. That cap matters when the mean is high. With a mean of 95, the mean plus one standard deviation could land above the best score anyone got.

```python
def compute_curve(df_0, min_grade=None):

    if min_grade:
        pass
    else:
        min_grade=np.max(df_0)/2.
    
    print("Min grade:",min_grade)

    df=df_0[df_0>min_grade]
    N_total=sum(df_0>min_grade)
    
    print("N Total (post cut):",N_total)
    print("N Total (pre cut):",df_0.shape)
    
    mean=np.mean(df)
    std=np.std(df)

    print("Mean:",mean)
    print("Std:",std)
    
    grade_def=[("A+",np.max(df)),
               ("A", min(np.max(df),mean+std)),
               ("B",mean), 
               ("C",mean-std),
               ("D",mean-2*std),
               ("F",0.)]
    
    grade_def.reverse()
    grades=OrderedDict( grade_def )
    
    
    hist,bins=np.histogram(df,bins=list(grades.values()))
    
    _=plt.hist(df)
    plt.show()
    
    for grade_letter,bin_count,min_grade in zip(grades.keys(),hist,grades.values()):
        print("{:2.2f}".format(min_grade),
              grade_letter+": "+str(bin_count),
              "{:2.2f}%".format(100.*float(bin_count/N_total)) )
    return grades
    
```

We have written a function to help us determine how to assign grades, but we haven't actually assigned any grades to any student. Here is a function that assign grades, given a dictionary of grade boundry definitions:

```python
def assign_grade(numeric,grades):
    previous_grade="F"
    for grade_letter,grade_boundry in grades.items():
        #print(numeric, grade_letter,grade_boundry)
        if numeric < grade_boundry:
            return previous_grade
            break
        else:
            previous_grade=grade_letter
    return previous_grade

    
```

## Grading Strategies

Thinking about what we did, we noticed that we have 2 grading strategies:
1. "Standard": 70, 80, 90 scheme.
2. "Curved": Using mean and standard deviation.

There is a third strategy that I often apply when the grade distribution doesn't look like a normal distribution. When we curve the letter grades mean the following, based on the fact that 68% of normal distribution lies between -1 and 1 $\sigma$ of the mean.

* B and C: comprise of the 68% of the students withing 1 $\sigma$ of the mean, with the top 34% subset assigned a B and the bottom 34% assigned a C. Nearly all of the remaining 32% of the students will be within 2 $\sigma$ of the mean.
* A: The approximately 16% that are more than 1 $\sigma$ above the mean are assigned an A.
* D: The approximately 16% that are more are between 1 $\sigma$ and 2 $\sigma$ below the mean are assigned a D.

All of this assume a normal distribution. If the grade distribution isn't normal, we can still assign grades in a way that we could interpret in the same way as the curved grades by simply assigning the top 16% an A, next 32% B, then assign C to the 32% and D to the the 16% after.

It's worth being clear about what a grade is for. Letter grades group students: these students did really well and are like each other, these did not do as well but are like each other, and so on. If the boundaries follow the groups you can see in the histogram, a student who finds themselves in the distribution can see why they got the grade they did.

Curving means mapping whatever distribution you actually have onto something you can interpret as a normal distribution. Even the uncurved 90/80/70 scale carries that assumption: it amounts to assuming the mean is at 80 with a standard deviation of 10. Used this way, the bell curve is the worst case and never a penalty: the grades given should be at least as good as a bell curve would give, unless that's impossible. Whatever the scheme, it has to treat everyone the same way. Two students with the same score must get the same letter, so a scheme is a set of lines drawn on the score axis, and it will come close to the target counts without hitting them exactly.

None of these schemes should be applied blindly. The process is to look at the data, think about your assumptions and what you're trying to achieve (is this fair to the students?), try the schemes, see their consequences and then decide. For every exam and every lab, the result of that analysis is a set of grade boundaries.




# Disecting the Task at Hand

We can also note that all the strategies are defined by the boundries between letter grades. Therefore we can factorize the operation of assigning a letter grade into several steps:
* Perform statistical analysis of the grades to obtain basic parameters.
    * Minimum cut-off
    * Mean / Sigma
* Establish grade boundries.
    * Potentially apply several strategies.
    * Select a strategy appropriate for given distribution of grades.
* Use grade boundries to assign grades.
* Validate the assigned grades.

This step is worth taking every time. Once you've worked through a problem by hand and realize you need to rewrite it properly, step back before writing any code and ask where the abstractions should be: what are the conceptual things I'm doing, and in what steps? Those steps are what the code should map onto. Two of them deserve a note. Selecting a strategy is done by hand, which means the system has to let you see what's going on so you can make that decision. And the final validation is there no matter what else you do, to check that the results are what you expected and nothing went wrong.



## Organizing Our Code

It is possible to perform the steps above without any (or with minimal) generalization or abstraction (ie using functions). We could work out a grade with a series of steps... then copy and paste for the other grades. The benefits would be that without generalization, we could fine tune every grade easily. But it could become tedious and our results would be error prone.

Alternatively, we can recognize that each of the steps above can be done with a set of functions that have the same inputs and outputs:

* Statistical Analysis Functions: Input list of grades. Output mean, standard deviation, and minimum grade.
* Grade Boundries Functions: Input stats (and list of grades). Output boundries.
* Assign Grades Function: Input boundries and list grades. Output corresponding list of assigned letter grades.
* Validate Function: Input assigned letter grades, output percentage of each grade.

In addition, we will need to do a weighted sum of the grades to assign a class grade. 

We also have to note that beyond the procedural steps, we will have to make choices about how we represent the input/output data of each step. For example, previously we represented grade boundries as an `OrderedDict`. 

Notice that each item on that list isn't a single function but a family of them. There will be several grading strategies, and because they have the same inputs and outputs, they look and operate the same way. That is exactly the kind of commonality object-oriented design can capture: each family can become one class of things. Looking at the steps, and at the inputs and outputs of each, is how you start to see how to organize the code.



## Building a System

We are starting to recognize that depending on the context, we may choose different levels of sophistication for our code to solve a problem:

* Quickly solve a problem once: Cut and paste can be sufficient.
* Scale or solve the problem on many data points: Implement functions that perform tasks.
* Solve the same problem with different data: Identify tasks and generalize functions with defined input/outputs
* Enable others to use your code to solve the same problem.
* Build a production system or service that solves the problem.


Each of these is a real step on a real project, not just a hypothetical progression. Picture it concretely: a client hands you a data file and wants an answer by tomorrow — you cut, paste, and hack until you have one, and that's a completely reasonable way to handle a one-off, narrow problem. Then they come back wanting the same thing done for a hundred more datasets — now cutting and pasting doesn't scale, so you write functions and loops around what you did by hand. Then it turns out the process needs to work on data that doesn't look quite like the original case, so you generalize what was hard-coded. Then someone else — a TA, a new hire — needs to use your code without having been in the room while you wrote it, which means documenting it, validating its inputs, and giving clear error messages instead of assuming they know what you know. And finally, the whole thing needs to run in production, at scale, for users who aren't you — which is a genuinely different problem from any of the ones before it, closer to what a software engineer does than what a data scientist does.

That last jump matters enough to say plainly: a big, mature piece of software accumulates *legacy* — design choices that made sense given what was known at the time, that later turn out to be limiting once the problem grows into something nobody originally anticipated. Rewriting from scratch is rarely the answer; what saves you is having kept your *interface* stable even as what's happening underneath gets redesigned, so the change doesn't ripple out to everyone using your code. That's the same idea we'll lean on below when we design these classes: separate the abstraction a caller sees from the implementation behind it, because the implementation is what you'll end up rewriting.

Two rungs of that ladder are worth filling in. Between handing your code to one person and building a production system there is often a middle stage. A whole group, say all the faculty in a program who want to grade the same way, keeps a shared repository of tools. They standardize it, design it with care and build it up collectively, the way a research team does, so that everyone can use each other's pieces. And when someone does want the production service, the first design decision is to separate the interface, what the user sees and interacts with, from the internal engine that does the computation, perhaps with a web server in between connecting the two. What we'll design here is the engine, and its design pattern is a common one for processing data well beyond grades.

Each stage takes a lot more effort than the one before, and what that effort buys is reliability, reproducibility and scalability. Along the way, you will often be on the receiving end: handed someone's code with no comments and no explanation, which crashes when you run it. Take that in stride; take your time, read it and figure it out.

## An Object Oriented Approach

Object Oriented programming is not required to solve problems. But in many cases, OO can greatly simplify the development process and ability to add further functionality to the code. Even within OO, there are many approaches to solving the same problem, and while some may have advantages, often it takes several iterations and revisions to approach optimality... and even then, there are stylistic and historical choices.


This isn't abstract: my own particle-physics experiment runs on roughly seven million lines of C++ developed over 25 years, and it still carries the marks of design choices made decades ago, because nobody designing it then could anticipate every way it would eventually be used. When you look at any large, long-lived system, there's a real sense of legacy and history behind why it looks the way it does — not because the original designers were careless, but because no one designs for a problem they haven't encountered yet.

Lets explore one OO approach to building a Grade Book system.

Before writing code, it helps to think about the design at a high level, with pictures. The standard notation for object-oriented design is the Unified Modeling Language (UML), which isn't tied to any one language. Each box is a class, listing its name, its attributes (the data it holds, and whether each is private or public) and its methods. An arrow from one box to another means inheritance. A diagram like this specifies who holds what data and how it is accessed, closely enough that someone could type out the skeleton of the code from it without any implementation. There are tools that do exactly that, generating every class's skeleton from the diagram and leaving you to fill in the functions. Drawn in UML, the `base` class below is a box that takes a name in its constructor, holds that name as its one attribute and has one accessor method to return it, and `data` and `alg` are boxes with arrows pointing to it.

## Base Classes

We are going to build some classes to represent our data and algorithms. Once everything is build, we will instantiate and configure these classes and load them up with data. Since we will be manilpulating many of them, we'll find it useful for all of our classes to have a name that is assigned at instantiation. One use of instance names will be when that if a class has to give a message back to us, it can identify itself by its name so we know who (which instance) it is.

Since we recognize that there may be common elements (e.g. name) that all of our classes will possess, let's begin for building base classes for everything. In addition, we have discussed the principle of separating data and algorithms. In order to explicitly enforce this separations, we branch our inheritance tree to explicitly separate the algorithms and data classes. We will have data classes that will hold information and allow us to manipulate that information in correct ways. We will also have algorithm classes that we configure in specific ways and then apply to the data classes.

It can seem strange to insist on keeping algorithms and data apart when the whole point of a class is to bundle code and data together — but that bundling is for building *abstractions* (a complicated class presenting a simple interface), not for gluing one particular algorithm to one particular data representation. Think of a PDF file: the format is the data, and there are many different readers — the algorithms — that can open it. If an algorithm and a data representation were fused into one class, comparing two different algorithms on the same data, or plugging in a new one later, would mean rewriting the data side too.

The instance name is for bookkeeping and debugging, not something shown to the outside world: with thousands of grades in memory, the one that hits an error can say which student's grade for which assignment it is. The name can also be used to look things up, the way the shapes in Lecture 9's drawing program had names so you could fetch a particular box, resize it and redraw it.

A base class that everything derives from is also the place for any feature you want every part of the system to have at once. Control over verbosity is a good example: sometimes you run a system interactively and want it chatty, and other times you want it completely quiet. If every object knew its verbosity level and checked it before printing a message, the whole system could be controlled globally, or object by object. It isn't implemented in this design, but the base class is where it would go.

The data/algorithm split also says what each side is allowed to do. The code in a data class exists only to make its information easy to retrieve and to keep it correct; it doesn't compute its own mean. Algorithm classes take data objects in and give data objects out, and hold nothing but configuration: a curve calculator configured with a minimum cutoff of 55 stores that 55, but the 55 isn't part of the data being processed. The system is then a set of modular components that pass data objects between them.

The `data` and `alg` classes below have no code in them yet. They're scaffolding, there so the two kinds of class can't be mixed up by mistake, and they pay off later. To make data persistent, for example, you would implement that once in `data`, and the whole data branch would have it, specialized where needed, while the algorithms, which don't need persisting, stay untouched.

```python
# Create some virtual classes

# Base Class for all other classes
# Holds instance name for run-time debugging
class base:
    def __init__(self,name):
        self.__name=name

    def name(self):
        return self.__name

# Data Classes
# These classes store data
# Methods facilitate accesss and ensure consistency of data
class data(base):
    def __init__(self,name):
        base.__init__(self,name)

# Algorithm Classes
# These classes encapsulate processes as module components
# The data in these classes hold configuration
# Methods input/output data classes
class alg(base):
    def __init__(self,name):
        base.__init__(self,name)
```

Example classes:
 * Data: Grade, Student, GradeBook
 * Algorithm: Compute Statistics, Determine Grade Boundaries, Assign Grades, Sum Grades, ...

## Data Representation

Our system will need to keep track of:
* Grades Associated with Assignments
* Students Associated with a class, each hold grades for various assignments.
* Gradebook corresponding to a set of students in a specific course.

How are these data types related? Lets ask ourselves some questions:
* Is there a hierachical relationship between the types of data? 
* Do the classes represent different implementations or specializations of a the same concepts?
    * Are Grades a type of Student or Gradebook?
    * If so then we should consider inheritance.
* Do the data types encapsulte each other?
    *  Students hold Grades
    *  Gradebook holds Students

Answering those questions: inheritance is the right relationship when classes are different implementations or specializations of the same concept, such as different kinds of grades, or different algorithms that all do the same job in different ways. A student isn't a specialization of a grade, and a grade book isn't a specialization of a student, at least not the way we're thinking of them. The relationship here is encapsulation: grades are held by students, students are held by the grade book, and the grade book is the top data object. Inheritance and encapsulation are the two basic ways data types relate. More complicated data structures may call for something else, such as objects that simply hold links to each other.

## Grade Representation

We note that we have two types of grades: numerical and letter. We have a choice of how to represent grades. Below we choose to create a single class for both numerical and letter grades that keeps track of the type of grade it is and behaves accordingly. Altneratively, we could create a grade base class and derive two different classes, one for numerical grades and another for letter grades. 

Neither choice is obviously correct, and there's no need to over-justify picking one: sometimes you go with the simpler option because it does what you need right now, without designing for a case you don't yet have. If a third kind of grade shows up later that doesn't fit, that's a reasonable time to go back and split the class — not before.

In the two-class version, the base grade class would hold a value and a *grade book name*, which is different from the instance name. The grade book name is what appears in a printout, like "Lab 2"; the instance name identifies this particular object, one student's Lab 2 grade. The `grade` class below keeps the same distinction. The single class chosen here takes on its personality at instantiation, numerical or letter, and keeps it. If a split turns out to be needed later, the interface can probably stay the same.

Over-design is a real failure mode. When scientific software moved to object-oriented C++ in the 1990s, some of it was badly over-designed: organized around analogies that didn't fit, bloated, and eventually rewritten. Design decisions also carry concrete costs. If a step you'll need to redo, such as a calibration, sits early in a processing chain, every change means reprocessing all the data after it, and for a large experiment that can take months. Sometimes it makes sense to think a design through and build something big; sometimes the simple solution is the right one. Here, it's the simple one.

```python
class grade(data):
    # These members are shared with all classes of the same type and doesn't belong to any specific instance.
    __letter_grades=["F-","F","F+","D-","D","D+","C-","C","C+","B-","B","B+","A-","A","A+"]
    
    def __init__(self,name,numerical=True,value=None):
        # These members belong to the instances of classes and not shared. 
        # Note all data members are private so we'll have to build accessors to change and see their values.
        self.__value=value
        self.__numerical=numerical
        self.__gradebook_name=str()
        
        if value:
            if isinstance(value,(int,float)):
                self.__numerical=True
            elif isinstance(value,str):
                self.__numerical=False
            self.set(value)
        else:            
            self.__numerical=numerical
            
        self.__gradebook_name=name
        data.__init__(self, name+" Grade Data Object")        

    # Setter: 
    def set(self,value):
        if isinstance(value,(int,float)) and self.__numerical:
            self.__value=value
        elif isinstance(value,str) and not self.__numerical:
            if value in self.__letter_grades:
                self.__value=value
        else:
            print( self.name()+" Error: Bad Grade.")
            raise Exception

    # Accessors
    def value(self):
        return self.__value
    
    def numerical(self):
        return self.__numerical
    
    def gradebook_name(self):
        return self.__gradebook_name
    
    def __str__(self):
        return self.__gradebook_name+": "+str(self.__value)
```

## Student Representation

Next type of data we need to represent is students:

```python
class student(data):
    
    def __init__(self, first_name, last_name, id_number):
        self.__grades=dict()
        self.__id_number=id_number
        self.__first_name, self.__last_name = first_name, last_name
        data.__init__(self,first_name+" "+last_name+" Student Data")

    def add_grade(self,a_grade,overwrite=False):
        if overwrite or not a_grade.gradebook_name() in self.__grades:
            self.__grades[a_grade.gradebook_name()]=a_grade
        else:
            print (self.name()+" Error Adding Grade "+a_grade.name()+". Grade already exists.")
            raise Exception

    def id_number(self):
        return self.__id_number
    
    def __getitem__(self,key):
        return self.__grades[key]
    
    def print_grades(self):
        for grade in self.__grades:
            print (self.__grades[grade])
```

## Algorithms

So far, we have put together data classes. Now lets design some algorithm classes that interact with the data.

When thinking about the different types of operations we performed in our notebook based data analysis, it became clear that we could classify them by their inputs and outputs:

* *Statistical Analysis Functions:* Inputs list of grades. Output: few statistical quantities.
    * Example: Input list of grades. Output mean, standard deviation, and minimum grade.
* *Grade Boundries Functions:* Input stats (and list of grades). Output boundries.
    * Example: Curve boundry calculation.
* *Assign Grades Function:* Input boundries and list grades. Output: Letter grade.
    * Example: Letter grade mapping.
* *Validate Function:* Input assigned letter grades, output percentage of each grade.

Furthermore, as we'll see in a bit, some of these operations have very similar logic (e.g. in terms of loops), with minor differences otherwise that we can easily encapsulate. We will implement these operations as `alg`s in our system. Since the input/outputs are different, we will have to think about how the classes implementing each of these operations relate to each other. Let consider two choices:

1. We create a separate common base class for all algorithms that have exactly the same input/outputs.
    * Each base class implements a different `apply_XXXX(...)` function with specific inputs as arguments and returning the outputs.
        * Some common patterns can be implemented these base classes.
    * Every operation inherits from appropriate base class and overrides `apply` method.
        * `apply` implementations generally simple since input are provided.
        * In some cases `apply` in the base class can implement the pattern, allowing for simpler `apply` in the derived class.
            * For example: all grade assigning algorithms will need to loop over all students and add an additional grade. We only need to write this part of the logic once and just code up different methods of computing the grade for each type of grading algorithm.
    * The caller of `apply` will need to identify the type of operation and supply the correct input.
1. We create a single base class with common a `apply(...)` that is provided the data objects as input.
    * The each operation's `apply` method retrieves inputs it needs directly from the data objects.
        * Minimal logic in the base class, everything done in the overridden `apply` method.

### Grade Calculator

We'll start with first approach... and later consider changing to the second. 

We begin by implementing the operation of using grade boundries to turn numerical grades into assigned letter grades. Here, the input is a numerical grade and the output is a letter grade. Our first example of such an algorithm computes letter grades using the usual 90, 80, 70 percent scheme.

```python
class grade_calculator(alg):    
    def __init__(self,name,stats):
        self.__stats=stats
        alg.__init__(self,name)

    def apply(self,a_grade):
        raise NotImplementedError
        

class uncurved_letter_grade_percent(grade_calculator):
    __grades_definition=[ (.97,"A+"),
                          (.93,"A"),
                          (.9,"A-"),
                          (.87,"B+"),
                          (.83,"B"),
                          (.8,"B-"),
                          (.77,"C+"),
                          (.73,"C"),
                          (.7,"C-"),
                          (.67,"C+"),
                          (.63,"C"),
                          (.6,"C-"),
                          (.57,"F+"),
                          (.53,"F"),
                          (0.,"F-")]
    
    def __init__(self,grade_name,max_grade=100.):
        self.__max_grade=max_grade
        self.__grade_name=grade_name
        grade_calculator.__init__(self,
                                  "Uncurved Percent Based Grade Calculator "+self.__grade_name+" Max="+str(self.__max_grade),
                                 False)
        

    def apply(self,a_grade):
        if not isinstance(a_grade,grade):
            print (self.name()+ " Error: Did not get an proper grade as input.")
            raise Exception
        if not a_grade.numerical():
            print (self.name()+ " Error: Did not get a numerical grade as input.")
            raise Exception

        percent=a_grade.value()/self.__max_grade
        
        for i,v in enumerate(self.__grades_definition):
            if percent>=v[0]:
                break
                            
        return grade(self.__grade_name,value=self.__grades_definition[i][1])
            
```

The calculator above refuses bad input by printing a message and then raising `Exception`. That's deliberately loose: `Exception` is the most generic kind of error, and Python has a whole tree of more specific ones, which we haven't covered yet. The idea for now is that in critical code you want the program to break when something goes wrong, rather than keep going and quietly hand you bad results, possibly without an error message you'd ever notice.

A curved grade can reuse the same table of boundaries. Instead of moving the boundaries, it shifts and scales the grade: convert it to a percentage, subtract the mean percentage so the distribution is centered at zero, divide by the standard deviation and multiply by 0.1 so the width is 10%, then add 0.8. That's the same shift-and-scale operation as in Lab 3. Afterwards the distribution has a mean of 80% and a standard deviation of 10%, and the letter grades come from the same table as before.

Now we can implement a curved grade:

```python
class curved_letter_grade(grade_calculator):
    __grades_definition=[ (.97,"A+"),
                          (.93,"A"),
                          (.9,"A-"),
                          (.87,"B+"),
                          (.83,"B"),
                          (.8,"B-"),
                          (.77,"C+"),
                          (.73,"C"),
                          (.7,"C-"),
                          (.67,"C+"),
                          (.63,"C"),
                          (.6,"C-"),
                          (.57,"F+"),
                          (.53,"F"),
                          (0.,"F-")]
    __max_grade=100.
    __grade_name=str()
    
    def __init__(self, grade_name, mean, std, max_grade=100.):
        self.__max_grade=max_grade
        self.__mean=mean
        self.__std=std
        self.__grade_name=grade_name
        grade_calculator.__init__(self,
                                  "Curved Percent Based Grade Calculator "+self.__grade_name+ \
                                  " Mean="+str(self.__mean)+\
                                  " STD="+str(self.__std)+\
                                  " Max="+str(self.__max_grade),
                                   False)
        

    def apply(self,a_grade):
        if not isinstance(a_grade,grade):
            print (self.name()+ " Error: Did not get an proper grade as input.")
            raise Exception
        if not a_grade.numerical():
            print (self.name()+ " Error: Did not get a numerical grade as input.")
            raise Exception
    
        # Rescale the grade
        percent=a_grade.value()/self.__max_grade
        shift_to_zero=percent-(self.__mean/self.__max_grade)
        scale_std=0.1*shift_to_zero/(self.__std/self.__max_grade)
        scaled_percent=scale_std+0.8
        
        for i,v in enumerate(self.__grades_definition):
            if scaled_percent>=v[0]:
                break
                            
        return grade(self.__grade_name,value=self.__grades_definition[i][1])
            
```

Note that the name of the grade of interest, the mean, and the standard deviation are provided in constructor of the algorithm.

### Stats Computation

Another operation we identified was computation of stats. Here the inputs are a list of grades, and output are a few summary statistics, for example the mean and standard deviation:

```python
import numpy as np
import math

class statistics_calculator(alg):    
    def __init__(self,name):
        alg.__init__(self,name)

    def apply(self,grades):
        raise NotImplementedError
        
class mean_std_calculator(statistics_calculator):
    def __init__(self):
        statistics_calculator.__init__(self,"Mean and Standard Deviation Calculator")
        
    def apply(self,grades):
        return np.mean(grades),math.sqrt(np.var(grades))

```

### Grade Summing

There is another type of operation we didn't think about earlier... summing grades. Here the input is a student (which is the object holding all of the grades), the output is a grade.

```python
class summary_calculator(alg):    
    def __init__(self,name):
        alg.__init__(self,name)

    def apply(self,a_student):
        raise NotImplementedError
        
class grade_summer(summary_calculator):
    def __init__(self,prefix,n):
        self.__prefix=prefix
        self.__n=n
        statistics_calculator.__init__(self,"Sum Grades")
        
    def apply(self,a_student):
        labels=[prefix+str(x) for x in range(1,n)]
        
        grade_sum=0.
        for label in labels:
            grade_sum+=a_student[label]
            
        a_student.add_grade(grade(prefix+"sum",value=grade_sum))
```

## Gradebook

Finally we need a data object to hold all of the students. 

```python
class grade_book(data):
    
    def __init__(self,name):
        data.__init__(self,name+" Course Grade Book")
        self.__students=dict()
    
    def add_student(self,a_student):
        self.__students[a_student.id_number()]=a_student
        
    def assign_grade(self,key,a_grade):
        the_student=None
        try:
            the_student=self.__students[key]
        except:
            for id in self.__students:
                if key == self.__students[id].name():
                    the_student=self.__students[id]
                    break
        if the_student:
            the_student.add_grade(a_grade)
        else:
            print (self.name()+" Error: Did not find student.")
            
    def apply_grader(self,a_grader,grade_name):
        for k,a_student in self.__students.items():
            a_student.add_grade(a_grader.apply(a_student[grade_name]))
            
    def apply_stats(self,a_stat_comp,grade_name):
        grades=list()
        for k,a_student in self.__students.items():
            grades.append(a_student[grade_name].value())
            
        return a_stat_comp.apply(grades)
        
    def print(self):
        for id,student in self.__students.items():
            print("------------------------------------")
            print("Student:",id)
            print("------------------------------------")
            student.print_grades()
            
    
```
