1. Write a "counter" class that can be incremented up to a specified maximum value, will print an error if an attempt is made to increment beyond that value, and allows reseting the counter. 

```python
class counter:
    def __init__(self,max_value,start_value,current_value):
        self.max_value = int(max_value)
        self.start_value = int(start_value)
        self.current_value = start_value
    
    def increment_value():
        n = int(input("Please provide amount for value increment:"))
        if n == 0  or n <= 0:
            raise ValueError("Increment amount must be positive integer.")
            
        elif n>0:
            current_value += n
            if start_value > max_value:
                raise ValueError("Maximum value exceeded.")
            else: 
                return current_value
    def reset_counter():
        current_value = start_value
        return current_value
    
        
```

```python
class rectangle:
    def __init__(self,l,w,x,y):
        self.__l = l
        self.__w = w
        self.__x = x
        self.__y = y
    def compute_area():
        area = self.__l * self.__w
        return area
    def compute_perimeter():
        perimeter = 2 * (self.__l + self.__w)
        return perimeter
    def get_attr():
        print("Length:",{self.__l},"/n",
              "Width:",{self.__w},"/n",
              "Coordinates:",{self.__x},{self.__y})
    
        
    
```

4. Implement a class to represent a circle, holding the radius and $x$ and $y$ coordinates of center of the object. Implement functions that compute the area and perimeter of the circle. Make all data members private and provide accessors to retrieve values of data members. 

```python
class circle:
    def __init__(self,r,x,y):
        self.__r = r
        self.__x = x
        self.__y = y
    def compute_area():
        area = (self.__r**2)*3.14
        return area
    def compute_circumference():
        circumference = 2 * 3.14 * self.__r
        return circumference
    def get_attr():
        print("Radius:",{self.__r},"/n",
              "Coordinates:",{self.__x},{self.__y})
```

12. Implement the ability to load/save raster drawings and demonstrate that your method works. One way to implement this ability:

   * Overload `__repr__` functions of all objects to return strings of the python code that would construct the object.

For example:

```python
class foo:
    def __init__(self,a,b=None):
        self.a=a
        self.b=b
        
    def __repr__(self):
        return "foo("+repr(self.a)+","+repr(self.b)+")"
    
    def save(self,filename):
        f=open(filename,"w")
        f.write(self.__repr__())
        f.close()
        
   
def foo_loader(filename):
    f=open(filename,"r")
    tmp=eval(f.read())
    f.close()
    return tmp
```
